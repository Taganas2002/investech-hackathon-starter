# Says vs does: management's M&A claims vs what the filings show

Each row: a claim from an earnings call (exact quote, checked word for word against the transcript: `quote_verified`), what happened next, and the filing evidence.
Extraction was AI-assisted and then checked against the source data. Topicus has no recent calls in the data, so it's not included.

```python
import pandas as pd
pd.set_option("display.max_colwidth", 200); pd.set_option("display.width", 220)
NAMES = {"csu": "Constellation", "toi": "Topicus", "tfii": "TFI", "wsp": "WSP", "aecon": "Aecon"}
cl = pd.read_csv("output/claims.csv")
cl["company"] = cl["key"].map(NAMES)
print(f"{len(cl)} claims, {cl['quote_verified'].sum()} quotes verified against transcripts")
cl.groupby(["company", "verdict"]).size().unstack(fill_value=0)
```

```python
# The full table (one company at a time reads best on screen)
COMPANY = "TFI"     # change to Constellation, WSP, Aecon
cl[cl["company"] == COMPANY][["call", "speaker", "claim_quote", "target", "outcome", "verdict", "check_by"]]
```

```python
# Demo view: one line per claim
def demo(company):
    for _, r in cl[cl["company"] == company].iterrows():
        print(f"[{r['call']}] {r['speaker']}")
        print(f'   SAID:  "{r["claim_quote"]}"')
        print(f"   DID:   {r['outcome']}")
        print(f"   → {r['verdict'].upper()}\n")
demo("TFI")
```
